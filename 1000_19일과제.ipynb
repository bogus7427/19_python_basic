```python
import re

# input()을 사용해서 텍스트 입력받기
text = input()

# \b를 이용해 독립된 소문자 'the'만 정확하게 찾기
matches = re.findall(r"\bthe\b", text)

# 찾은 개수 출력
print(len(matches))
```

```python
# 1. 입력받은 문자열을 ';' 기준으로 쪼갠 후 정수(int)로 변환하여 리스트로 저장
prices = list(map(int, input().split(";")))

# 2. 높은 가격순(내림차순)으로 정렬
prices.sort(reverse=True)

# 3. 각 가격을 길이 9, 오른쪽 정렬, 천 단위 콤마 형식으로 출력
for p in prices:
  print(f"{p:>9,}")
```
