```python
cal, row = map(int, input().split())

matrix = []
for i in range(row):
    matrix.append(list(input()))
for i in range(row):
    for j in range(cal):
        if matrix[i][j] == '*':
            print('*', end='')
        else:
            count = 0
            for di in range(-1, 2):
                for dj in range(-1, 2):
                    ni = i + di
                    nj = j + dj
                    if 0 <= ni < row and 0 <= nj < cal:
                        if matrix[ni][nj] == '*':
                            count += 1

            print(count, end='')
    print()
```

```python
cal, row = map(int, input().split())  # 가로, 세로 입력

matrix = []  # 2차원 리스트 생성

for i in range(row):  # 세로 줄 수만큼 반복
    matrix.append(list(input()))  # 각 줄 입력

for i in range(row):  # 각 행 확인
    for j in range(cal):  # 각 열 확인
        if matrix[i][j] == '*':  # 현재 칸이 지뢰이면
            print('*', end='')  # 지뢰 출력
        else:
            count = 0  # 주변 지뢰 개수 초기화

            for di in range(-1, 2):  # 위, 가운데, 아래
                for dj in range(-1, 2):  # 왼쪽, 가운데, 오른쪽
                    ni = i + di  # 주변 행 번호
                    nj = j + dj  # 주변 열 번호

                    if 0 <= ni < row and 0 <= nj < cal:  # 범위 확인
                        if matrix[ni][nj] == '*':  # 주변에 지뢰가 있으면
                            count += 1  # 개수 증가

            print(count, end='')  # 지뢰 개수 출력
    print()  # 다음 줄로 이동
```
