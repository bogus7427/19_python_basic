```python
#위치인수 

def print_numbers(a,b,c):
    print(a)
    print(b)
    print(c)
```

```python
print_numbers(10,20,30)
```

```python
#언팩킹 사용하기  *x  = [10,20,30]  같다
x= [10,20,30]
print_numbers(*x)
```

```python
#가변인수 함수 만들기
# #def 함수이름(*매개변수):
#    코드

def print_numbers(*args):
    #print(args)
    for arg in args:
        print(arg)
```

```python
print_numbers(10,20,30,40)
```

```python
#키워드 인수

def personal_info(name, age, address):
    print("이름: ",name)
    print("나이: ",age)
    print("주소: ",address)
    
```

```python
personal_info("홍길동","30","서울시용상구")
```

```python
personal_info(age=30, address="서울시", name="홍길동")

```

```python
def personal_info(**kwargs):
    #print(kwargs)
    for kw,arg in kwargs.items():
        print(kw,arg)
```

```python
personal_info(name="홍길동",age=30, address="서울시 용산구")
```
