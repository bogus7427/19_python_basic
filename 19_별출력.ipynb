```python
#중첩 루프 사용하기
for i in range(5):
    for j in range(5):
        print("j:", j, sep="", end=" " )
    print("i:",i,"\\n", sep="")
```

```python
#사각형으로 별 출력하기
for i in range(5):
    for j in range(5):
        print("*", end="")
    print()
```

```python
#사각형 모양 바꾸기
for i in range(3):
    for j in range(7):
        print("*", end="")
    print()
```

```python
#계단식으로 별 출력하기
for i in range(5):
    for j in range(5):
        if j  <= i :
            print("8", end="")
    print()
```

```python
#대각선 별 출력하기    #print()  아무것도 적지않는 기본값은 줄 바꿈 
for i in range(5):
    for j in range(5):
        if j == i:
            print("*", end=" ")
        else:
            print(" ", end=" ")
    print()
```

```python
for i in range(5):
    for j in range(5):
        if j < i:
            print(" ", end="")
        else:
            print("*", end="")
    print()
```

```python
for i in range(5):
    for j in range(7):
        if j == i:
            print(" " * (4 - i) + "*" * (2 * i + 1))
    print()
```

```python
n=5

for i in range(n):
    print(" " * (n-i-1)+"*"*(2*i+1))

```
