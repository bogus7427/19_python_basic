```python
students = [
    {'이름': '김민수', '국어': 85, '영어': 90, '수학': 78},
    {'이름': '이영희', '국어': 92, '영어': 88, '수학': 95},
    {'이름': '박철수', '국어': 78, '영어': 85, '수학': 82}
]
print("=== 학생 성적 관리 ===")

for student in students:
    total = student['국어'] + student['영어'] + student['수학']
    average = total / 3

    student['총점'] = total
    student['평균'] = average

    print("이름:", student['이름'])
    print(f"국어: {student['국어']}점, 영어: {student['영어']}점, 수학: {student['수학']}점")
    print(f"총점: {student['총점']}점, 평균: {student['평균']:.2f}점")
    print("---")
```

```python
text = "python is great python is fun python is powerful"
words = text.split()
word_count = { }
for word in words:
    if word in word_count:
        word_count[word] += 1
    else:
        word_count[word] = 1
for word, count in word_count.items():
    print(f"{word}: {count}번")
```

```python
cal, row = map(int, input().split())  # 가로, 세로 입력

matrix = []  # 2차원 리스트 생성

for i in range(row):  # 세로 줄 수만큼 반복
    matrix.append(list(input()))  # 각 줄 입력

for i in range(row):  # 각 행 확인
    for j in range(cal):  # 각 열 확인
        if matrix[i][j] == '*':  # 현재 칸이 지뢰이면
            print('*', end='')  # 지뢰 출력
        else:
            count = 0  # 주변 지뢰 개수 초기화

            for di in range(-1, 2):  # 위, 가운데, 아래
                for dj in range(-1, 2):  # 왼쪽, 가운데, 오른쪽
                    ni = i + di  # 주변 행 번호
                    nj = j + dj  # 주변 열 번호

                    if 0 <= ni < row and 0 <= nj < cal:  # 범위 확인
                        if matrix[ni][nj] == '*':  # 주변에 지뢰가 있으면
                            count += 1  # 개수 증가

            print(count, end='')  # 지뢰 개수 출력
    print()  # 다음 줄로 이동
```
