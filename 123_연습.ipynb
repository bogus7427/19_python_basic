```python
# 1. 함수 만들기 (정의)
def add_numbers(a, b):
    result = a + b
    return result

# 2. 함수 사용하기 (호출)
sum_result = add_numbers(3, 5)
print(f"3과 5를 더한 결과는: {sum_result} 입니다.")
```
