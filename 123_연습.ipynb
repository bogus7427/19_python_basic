```python
# 1. 함수 만들기 (정의)
def add_numbers(a, b):
    result = a + b
    return result

# 2. 함수 사용하기 (호출)
sum_result = add_numbers(3, 5)
print(f"3과 5를 더한 결과는: {sum_result} 입니다.")
```

```python
# 1. map()의 첫 번째 인자(int)와 두 번째 인자(input().split())를 콤마로 정확히 분리합니다.
a, b, c, d = map(int, input("숫자 4개를 입력하세요: ").split())

# 2. 평균 구하기
print((a + b + c + d) / 4)
```
