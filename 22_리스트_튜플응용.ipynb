```python
#while 반복문으로 요소 출력하기

a=[38,21,53,62,19]

i=0
while i < len(a):
    print(a[i])
    i +=1
    
```

```python
#가장 작은수와 가장 큰수 구하기

a=[38,21,53,62,19]

smallest =a[0]
for i in a:
    if i < smallest:
        smallest = i
print(smallest)
```
